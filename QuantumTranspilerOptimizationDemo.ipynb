```python
from qiskit import QuantumCircuit, transpile
from qiskit.providers.basic_provider import BasicSimulator


#Create an example circuit is like a simplified Grover
qc = QuantumCircuit(3, name="example_circuit")
qc.h([0, 1, 2])
qc.cz(0, 2)
qc.x(1)
qc.h([0, 1, 2])

print("Original circuit:")
print(qc)
print("Depth:", qc.depth())
print("Gate counts:", qc.count_ops())

 #Choose fake backend that represents real hardware
backend = BasicSimulator()


#Try several optimization levels 4

for level in range(4):
    print(f"\n=== Optimization level {level} ===")
    tqc = transpile(qc, backend=backend, optimization_level=level)
    print(tqc)
    print("Depth:", tqc.depth())
    print("Gate counts:", tqc.count_ops())
```
